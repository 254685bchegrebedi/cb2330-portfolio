$$ \text{Dot product}= \sum_{i=1}^{N} a_i b_i $$
The dot product multiplies corresponding elements of two vectors of equal length and then adds all the products together, measuring the degree of weighted combination of the two vectors.

```python
def dot_loop(a, b):
  total = 0
  for ai, bi in zip(a, b):
    total = total + ai*bi
  return total

vec_a = [1, 2, 3, 4]
vec_b = [5, 6, 7, 8]

res_loop = dot_loop(vec_a, vec_b)
print("dot product =", res_loop)
```

```python
import numpy as np
res_np = np.dot(vec_a, vec_b)
print("dot product =", res_np) #70
```

```python
print(f"Loop result: {res_loop}")
print(f"Numpy result: {res_np}")
print(f"Are equal? {res_loop == res_np}")
```
